# 03 - KPI Design
## Global Superstore Dataset (2011-2014)

**7 KPIs مبنية مباشرة على نتائج الـ EDA**، مش KPIs مفروضة مسبقًا.

```python
import pandas as pd
import sys
sys.path.insert(0, '../src')
from kpi_calculations import (
    total_revenue_and_profit, overall_profit_margin, average_order_value,
    profit_margin_by_group, high_discount_impact, sales_cagr, kpi_summary
)

df = pd.read_csv('../data/processed/cleaned_superstore.csv', parse_dates=['Order Date', 'Ship Date'])
df.head()
```

## KPI 1: Total Revenue & Total Profit

```python
rev_profit = total_revenue_and_profit(df)
print(f"Total Revenue: ${rev_profit['total_revenue']:,.0f}")
print(f"Total Profit:  ${rev_profit['total_profit']:,.0f}")
```

## KPI 2: Overall Profit Margin

```python
margin = overall_profit_margin(df)
print(f"Overall Profit Margin: {margin:.2f}%")
```

## KPI 3: Average Order Value (AOV)

**تنبيه مهم:** لازم نحسبه على مستوى الطلب (Order ID) مش على مستوى السطر (Line Item)،
لأن الطلب الواحد ممكن يحتوي أكتر من منتج.

```python
aov = average_order_value(df)
print(f"AOV (mean, per order):   ${aov['aov_mean']:.2f}")
print(f"AOV (median, per order): ${aov['aov_median']:.2f}")

# للمقارنة: الطريقة الغلط الشائعة
wrong_aov = df['Sales'].mean()
print(f"\nWrong AOV (line-item average): ${wrong_aov:.2f}  <- غلط، بيقلل القيمة الحقيقية")
```

## KPI 4: Profit Margin by Category

```python
category_margin = profit_margin_by_group(df, 'Category')
print(category_margin.round(2))
```

## KPI 5: Profit Margin by Market

```python
market_margin = profit_margin_by_group(df, 'Market')
print(market_margin.round(2))
```

## KPI 6: High-Discount Order Impact

الحجم المالي الفعلي لمشكلة الخصومات فوق 20%.

```python
impact = high_discount_impact(df, threshold=0.20)
for k, v in impact.items():
    print(f'{k}: {v}')
```

## KPI 7: Sales CAGR (Compound Annual Growth Rate)

```python
cagr = sales_cagr(df)
print(f"Sales CAGR (2011-2014): {cagr:.1f}% per year")
```

## Full KPI Summary (Dashboard-ready)

```python
import json
summary = kpi_summary(df)
print(json.dumps(summary, indent=2, default=str))
```
